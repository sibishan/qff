## Branching (Making Decisions)

Now that we can take input, store and process values, and output results,
what happens if we want to do something different depending on the state
of our program? Since we are now able to take in user input,
and that can vary a lot, we may want to do different things depending
on the values we have stored in our variables.

To achieve this, we have a tool called **branching**,
which allows our program to decide to take one action instead of another
all on its own, without us having to go in and manually change its
behaviour every time.

Generally, the logical structure of a branch in a program will look like:
```
if <condition 1>:
    <do action 1>
else:
    <do alternate action>
```

The conditions are where Boolean (true/false) values come very handy,
since these values are used to drive the decision-making of our program.
We already deal with such values in real life,
and you can think of Boolean values as **answers** to a yes/no question.

### Useful Tool: Logical Operators

A common set of operators you'll end up using
as conditions for your branches are the traditional
comparison operators you would have seen in maths before.
These are written as follows
(where `a` and `b` are the variables or values you wish to test):

* `a == b` tests if `a` is equal to `b`.
* `a != b` tests if `a` is not equal to `b`.
* `a > b` tests if `a` is strictly greater than `b`.
* `a >= b` tests if `a` is greater than or equal to `b`.
* `a < b` tests if `a` is strictly less than `b`.
* `a <= b` tests if `a` is less than or equal to `b`.

### Example 1: Basic If-Clause (One Case)

Here's an example program that takes in an integer
from the user, and prints out "six!"
if the user inputted a `6`. Otherwise, nothing happens.

Take note that if you want code to be executed
inside the branch of a particular condition
(much like the `print('six!')` line),
you have to **indent** your code.

Notice how there are 4 spaces at the front of
the print statement. This is the indent that tells Python
to only execute that print statement if the condition
`x == 6` evaluates to True.

````python
x = int(input('Enter an integer here: '))
if x == 6:
    print('six!')
````

### Example 2: If-Else Blocks (Two Cases)

Sometimes, you'd want to do something else if the condition
does not evaluate to True. Your logic would thus follow the structure:
"If `<condition>`, then `<action 1>`. Otherwise, `<action 2>`".

This language directly translates to the structure of an if-else block.

In this example, we output "less than 7" if a user's inputted integer
is less than `7`, otherwise, we output "at least 7".

````python
x = int(input('Enter an integer here: '))
if x < 7:
    print('less than 7')
else:
    print('at least 7')
````

Notice how we can do the exact same thing with the condition inverted
using the following (feel free to verify this with different values):

````python
x = int(input('Enter an integer here: '))
if x >= 7:
    print('at least 7')
else:
    print('less than 7')
````

### Example 3: If-Elif-Else Blocks (Multiple Cases)

But what happens if you want more?
There could be several conditions you'd want to test
before finally reaching an "everything else" case.
It is theoretically possible to do that by just nesting
more and more if-else blocks within the `else` clause
of an existing block, but your indents will get larger
and larger and quickly become unreadable
(see the example after this for details).

That's why there is a handy tool to augment our
`if-else` blocks. You can add `elif` blocks in between
the first `if` and the last `else` to check for
additional cases before falling back onto the final possibility.

As an example, let's imagine we have a LEGO set suitable
for the age ranges 3-99. If you're below 3, you're too young,
and if you're above 99, you're too old.
We can write the logic for this as follows.

````python
age = int(input('Input your age here: '))
if age < 3:
    print("You're too young!")
elif age > 99:
    print("You're too old!")
else:
    print('You can have this LEGO set!')
````

### Example 4: Nested If-Else Blocks

Of course, even though we have access to both the `if-else`
and the `if-elif-else` pattern to extend our logic further,
there are times where we really do need to nest code blocks.
Just remember to add indentation to refer to inner blocks,
as demonstrated in the following example.

````python
number = int(input('Enter an integer: '))
if number > 0:
    print('Positive')
    # Inner if-else block nested inside the outer one.
    if number % 2 == 0:
        # Notice how this line is now *two* indents deep instead of just one.
        print('Positive and even')
    else:
        print('Positive and odd')
elif number < 0:
    print('Negative')
````

## Looping (Repeating Certain Procedures)

Aside from the program needing to be able to make decisions on its own,
it may sometimes need to repeat certain commands to get something done.
Each time the body of a loop is executed once is called an **iteration**.
Again, you may not always know exactly how many times you need to
repeat something to accomplish your task, so this is yet another
fundamental functionality that a program needs to have.

There are two main types of loops Python has:
* A **for**-loop which repeats a set number of times
(the number of times can be determined by a variable,
and doesn't have to be hard-coded), and
* A **while**-loop which repeats until a certain condition is satisfied.
This is useful because sometimes,
even the program state cannot definitively tell you exactly how many
iterations are needed.

### Example 1: For-Loop (Basic Counting Loop)

We'll show examples of for-loops here.
The first kind is a basic one that has the exact number of iterations
coded directly into it, showing the basic syntax of a for-loop.

The second kind shows how you can use variables and also other
if-else statements in tandem with a for-loop.

````python
# Let's start with summing up 5 integers that are inputted by the user.

# We need to accumulate the sum of our integers into a designated variable for us to reference consistently.
# Let's call this variable "total".
total = 0

# Here's the main for loop.
# The variable `i` will take values 0, 1, 2, 3, and then 4, yielding five different iterations.
for i in range(5):
    # Additional tool here: f-strings are when you put "f" in front of a string literal,
    # allowing you to embed the values of existing variables using braces directly into the string.
    x = int(input(f'Enter number {i + 1}: '))
    # Notice the number `1` was added to `i` to make the counting convention more natural (1, 2, 3, 4, 5).

    # Now, I update our "total" to have the value of the current `x` added to it.
    total = total + x

print('Your total is:', total)
````

In the above code snippet, you can read the code and immediately see that
it will loop exactly five times. But it doesn't have to be like that!

Now we'll make a program that loops from the number `1` up until
and including a user-inputted integer value.
At every iteration, it outputs "Fizz" if the number is divisible by 3,
"Buzz" if the number is divisible by 5,
or "FizzBuzz" if both are true.
If none of those are true, then the number itself is outputted.

````python
upper_bound = int(input('Enter upper limit here: '))

# Given a `range(a, b)`, Python will loop starting from `a`, up until but not including `b`.
# Adding `1` to `b` makes the loop include `b` as its last iteration (otherwise it will be `b - 1`).
for num in range(1, upper_bound + 1):
    # Remember, indentation is important!
    if num % 15 == 0:
        # If divisible by both 3 and 5, then it is divisible by 15.
        # Notice how I test this *first*, since it is the stricter condition.
        print('FizzBuzz')
    elif num % 3 == 0:
        print('Fizz')
    elif num % 5 == 0:
        print('Buzz')
    else:
        print(num)
````

### Example 2: While Loop

Sometimes, no matter what you do, the program will never know for sure
exactly how many times to loop. Instead, the program needs to conduct a
check at the end of every iteration to see if a condition is no longer
satisfied.

An example where this is particularly useful
is when trying to take in user input,
but only terminating once a certain special value is entered.
In the following example, we keep outputting the user's value plus one,
until the user enters the special value of -1.

````python
num = int(input('Enter a value (-1 to stop): '))
while num != -1:
    print('Here is your number incremented by 1:', num + 1)
    num = int(input('Enter a value (-1 to stop): '))
````

## Lists

Sometimes, much like loops that could go on indefinitely,
you'll need to store many values at once while being able to
refer to them in a systematic way under the same variable name.
Turns out, you can! Aside from basic types like integers or decimals,
you can have a single variable be a *list* of values.

Lists are written as a comma-delimited sequence of values,
enclosed by square brackets. For example, a list with values 3, 4, and 5
is written as `[3, 4, 5]`.
This list will have a length of 3,
and you can access each element using an index.

One tricky thing with indices:
a computer starts counting from **0** instead of 1,
meaning you'd get the first element using the index `[0]` instead of `[1]`.
This is elaborated in the first example below.

### Example 1: Hard-Coded List and Interaction with For-Loops

Let's first put our list `[3, 4, 5]` into a variable we can access,
and start accessing and modifying its contents.

First, we'll loop over its contents via indexing,
then edit the second value, and output its contents again.

````python
first_list = [3, 4, 5]

# The `len()` function returns the length of the list (the number of elements in it).
print("The list's length is", len(first_list))

# We loop through, also printing out each of the indices used for your reference.
for i in range(len(first_list)):
    print('At index', i, 'the value is', first_list[i])

# Now we edit the second element. Notice the index used is 1 instead of 2.
first_list[1] = 1

# We loop through again, showing the changes.
for i in range(len(first_list)):
    print('At index', i, 'the value is', first_list[i])
````

### Example 2: Storing Many Inputs and Calculating a Value

The previous example was easy enough in the sense that we knew the length
of the list to begin with, and we never changed that length.
But if we want to add elements one by one into a list,
we're going to need a new method besides just indexing.
This is where the `.append()` method comes in.

Let's see this in action in tandem with a while-loop
for building the list, followed by a for-loop to calculate
the average value of the list.

````python
# You can initialise an empty list with just square brackets and nothing else in them.
numbers = []

input_num = int(input('Input number here (-1 to stop): '))
while input_num != -1:
    # The syntax for appending to a list is `<list name>.append(<value you want to add>)`
    numbers.append(input_num)
    # Grab the next integer value.
    input_num = int(input('Input number here (-1 to stop): '))

total = 0
for i in range(len(numbers)):
    # This time, let's use the short-hand for adding to a variable.
    # See the "Variables" section earlier for a previous example.
    total += numbers[i]

average = total / len(numbers)
print('The average of the list is', average)
````

### Example 3: More Looping Over Lists

There is also a more convenient way to iterate over lists.
You can do away with needing the indices of the lists to reference
the elements, and instead you can just iterate over each element directly.

````python
# Here's another example list.
pi_list = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

# Here, you have a variable `digit` that gets assigned the value of every element in the list `pi_list`.
for digit in pi_list:
    print(digit)

# Note, this is the same as doing:
# for i in range(len(pi_list)):
#     print(pi_list[i])
````

## Functions

Functions are tools that help define particularly important pieces of logic
that you need for your code, and allow you to easily reuse that logic
in multiple applicable points within your program.

In a sense, functions work a little like miniature versions of a program
that you'd write. It takes in some values (called **parameters**),
does some processing, and returns back a result.
The only difference is that you don't have to
take values via `input()` and output via `print()`.

You can place any kind of logic within the function,
whether it be storing local variables, branching, loops,
or even printing in the middle.
The neat thing about functions is that whatever variables you create
inside it are not accessible in the outer parts of the program
after the function completes execution, which means that you can name your
variables whatever you want without accidentally modifying something
on the outside!

You've actually been using some functions already
(`print()`, `input()`, `int()`, and `len()` to name a few).

We'll go through a couple of examples to show you how to define a function.

### Example 1: Basic Function Definition

Let's start with just a basic function that takes in nothing
and prints out a generic "Hello" string.
Note that this function takes no input parameters
and outputs nothing directly (it just prints to the screen).

When you want to *use* the function you've defined,
that's also known as **calling** the function.
That is also shown in the code snippet below, after defining the function.

````python
# This is the function definition...
def basic_function():
    print('Hello')

# ...and this is the function call.
basic_function()
````

### Example 2: Taking Input and Returning a Value

Next, let's make a function that takes in a single integer value
and outputs a list of all numbers from `1` up until and including that value.

````python
def number_list(n):
    output_list = []
    for num in range(1, n+1):
        output_list.append(num)
    # `return` is the keyword for providing output from a function, rather than printing.
    return output_list

# Now let's use it.
print(number_list(5))
````

### Example 3: Taking Multiple Inputs

Let's take three numbers, multiply the first two,
and add the third one.
Note that when you define the parameters of the function,
the order matters.

````python
# Three inputs are taken in a list: (a, b, c)
def mul_add(a, b, c):
    return a * b + c

# Notice how the order matters and affects the output in the two calls below.
print('4 * 2 + 3 =', mul_add(4, 2, 3))
print('4 * 3 + 2 =', mul_add(4, 3, 2))
````

### Example 4: Outputting Multiple Values (Tuples)

You can actually output multiple values at once using Python **tuples**.
Tuples are basically just bundles of data, with the main difference from lists
being that you can't grow them in size. You can still index them the same way.

````python
def add_and_sub(a, b):
    add = a + b
    sub = a - b
    return add, sub

# You can take the tuple and unpack its elements like so:
added, subtracted = add_and_sub(5, 2)
print('Added:', added, 'Subtracted:', subtracted)

# You can alternatively grab them by indices:
pair = add_and_sub(5, 2)
print('First:', pair[0], 'Second:', pair[1])
````

## Classes

Finally, the last major language feature of Python that you'll run into
potentially often during this hackathon is that of **classes**.
Classes are templates for **objects**, which we have dealt with already.
In fact, integers, decimals, strings, and even lists are all objects.

Objects in Python are models of objects in real life:
* They have some sort of internal state
represented by **attributes** which behave like local variables, and
* They have a set of operations you can conduct on them
represented by **methods** which behave like functions,
and can either return you some values based on the object's state
or even modify the object's state.

Knowing how to use methods attached to objects will be useful
when you begin using **Qiskit** for your quantum game.

We'll show a basic class definition below.

````python
# Here's a relatively trivial class for now (containing a pair of integers and some operations you can conduct on them).
class IntegerPair:
    # This is the constructor which is a special method: it is called when you *create* a new instance of this class (i.e., an object).
    def __init__(self, first, second):
        # So far, we just store the two numbers required by the constructor.
        self.first = first
        self.second = second

    # A method that returns the value of the first integer.
    def get_first(self):
        return self.first
    # A method that returns the value of the second integer.
    def get_second(self):
        return self.second

    # A method that sets the value of the first integer to the parameter you provide.
    def set_first(self, new_first):
        self.first = new_first
    # A method that sets the value of the second integer to the parameter you provide.
    def set_second(self, new_second):
        self.second = new_second

    # Element-wise adds the first and second integers of this IntegerPair instance and another instance, creating a new IntegerPair object.
    def add(self, other):
        # Here, the constructor (defined as `__init__`) is called.
        return IntegerPair(self.first + other.first, self.second + other.second)


# Let's try using it.
p1 = IntegerPair(3, 4)
p2 = IntegerPair(4, 5)

p3 = p1.add(p2)
print('P3: First:', p3.get_first(), 'Second:', p3.get_second())

p1.set_first(1)
p1.set_second(2)
print('P1: First:', p1.get_first(), 'Second:', p1.get_second())
````

Next, let's actually use Qiskit in our Python program!

We'll need to do the following:
* **Import** the Qiskit module (and Qiskit Aer module for simulating running the circuit).
* Create a new Quantum Circuit (we're making a new **object** here).
* Apply some quantum gates to the circuit,
as well as measurements at the end (we'll use **methods** of the object).
* Run the quantum circuit (using Qiskit Aer), and view the results.

The following example will generate a random $n$-length bitstring uniformly
using just Hadamard gates.

$n$ will be an integer inputted by the user,
and the number of times the circuit will be run is inputted as well.

````python
# For making quantum circuits in Qiskit.
import qiskit
# For running the quantum circuits without access to proper quantum hardware.
import qiskit_aer

n = int(input('How long will the bitstring be? '))
times = int(input('How many times will you run the circuit? '))

# `QuantumCircuit` is a class defined in the `qiskit` module.
# The two parameters are the number of qubits and number of classical channels respectively.
qc = qiskit.QuantumCircuit(n, n)

for i in range(n):
    # The method `.h()` applied a Hadamard gate to the (i+1)-th qubit.
    # The indexing convention is zero-based, just like Python lists.
    qc.h(i)

for i in range(n):
    # Measures the (i+1)-th qubit onto the (i+1)-th classical bit.
    qc.measure(i, i)

# The quantum circuit `qc` is built! Now we'll run it once.
backend = qiskit_aer.AerSimulator()

# `shots` is a keyword argument to the function (`shots` is not a new variable, but rather the name of the parameter being inputted).
# The value of the variable `times` is being passed to the `shots` parameter.
job = backend.run(qc, shots=times)

result = job.result()

# `counts` is a Python dictionary, so the order of the recorded bitstrings isn't guaranteed.
counts = result.get_counts()

for bitstring, count in counts.items():
    print('Bitstring', bitstring, 'was measured', count, 'times')
````
